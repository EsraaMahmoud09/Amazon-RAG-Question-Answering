```python
# =============================================================================
# Standard Library
# =============================================================================
import re
import string
from nltk.corpus import stopwords

# =============================================================================
# Data Processing
# =============================================================================
import pandas as pd
import numpy as np
```

```python
# =============================================================================
# Load the processed dataset
# =============================================================================
dataset = pd.read_csv("cleaned_dataset.csv")
```

```python
# Semantic Text Preprocessing
# =============================================================================
# Clean text for semantic embedding models
# =============================================================================

def clean_text_semantic(text):
    """
    Clean text while preserving semantic meaning.
    """

    # Convert to string
    text = str(text)

    # Convert text to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", " ", text)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text


# =============================================================================
# Create semantic text for embedding generation
# =============================================================================

dataset["semantic_text"] = (
    "Question: "
    + dataset["QuestionText"].apply(clean_text_semantic)
    + "\n\nAnswer: "
    + dataset["AnswerText"].apply(clean_text_semantic))

# Preview semantic text
dataset[["semantic_text"]].head()
```

```python
# Lexical (BM25) Text Preprocessing
# =============================================================================
# Clean text for BM25 lexical retrieval
# =============================================================================

def clean_text_bm25(text):
    """
    Clean text for lexical search.
    """

    # Convert to string
    text = str(text)

    # Convert text to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", " ", text)

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

```python
# BM25 Tokenization
# =============================================================================
# Load English stopwords
# =============================================================================

stop_words = set(stopwords.words("english"))


# =============================================================================
# Prepare text for BM25 indexing
# =============================================================================

def preprocess_bm25(text):
    """
    Clean, tokenize, and remove stopwords for BM25.
    """

    # Apply lexical cleaning
    text = clean_text_bm25(text)

    # Split into tokens
    tokens = text.split()

    # Remove stopwords
    tokens = [
        token
        for token in tokens
        if token not in stop_words]

    # Join tokens into a single string
    return " ".join(tokens)
```

```python
# Create BM25 Text
# =============================================================================
# Create lexical text for BM25 retrieval
# =============================================================================

dataset["lexical_text"] = (
    dataset["QuestionText"].apply(preprocess_bm25)
    + " "
    + dataset["AnswerText"].apply(preprocess_bm25))

# Preview lexical text
dataset[["lexical_text"]].head()
```

```python
# =============================================================================
# Compare original and processed text
# =============================================================================

sample = 0

print("Original Question")
print(dataset.loc[sample, "QuestionText"])

print("\nOriginal Answer")
print(dataset.loc[sample, "AnswerText"])

print("\nSemantic Text")
print(dataset.loc[sample, "semantic_text"])

print("\nLexical Text")
print(dataset.loc[sample, "lexical_text"])
```

```python
# =============================================================================
# Save the preprocessed dataset
# =============================================================================
dataset.to_csv("preprocessed_dataset.csv", index=False)

print("Preprocessed dataset saved successfully.")
```
